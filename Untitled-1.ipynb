```python
from safetensors import safe_open
import jax.numpy as jnp

weights = {}
for shard_path in sorted(checkpoint_dir.glob("*.safetensors")):
    with safe_open(shard_path, framework="np") as f:
        for name in f.keys():
            weights[name] = jnp.asarray(f.get_tensor(name))
```

```python
import json
from dataclasses import dataclass
from pathlib import Path
from typing import Any

import jax
import jax.numpy as jnp
import treescope
from safetensors import safe_open

def nest_weights(weights: dict[str, jax.Array]) -> dict[str, Any]:
    """Turn HF's dotted flat names into a nested dict so treescope.show() renders a real tree."""
    nested: dict[str, Any] = {}
    for name, tensor in weights.items():
        node = nested
        *path, leaf = name.split(".")
        for key in path:
            node = node.setdefault(key, {})
        node[leaf] = tensor
    return nested
```

```python
treescope.show(nest_weights(weights)) 
```
